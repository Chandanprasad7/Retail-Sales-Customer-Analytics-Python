```python
import pandas as pd

file_path = "../data/Online Retail.xlsx"

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

df.head()
```

```python
print("DATASET INFORMATION")
print("-------------------")

df.info()

print("\nDATA TYPES")
print("----------")
print(df.dtypes)

print("\nDATASET SIZE")
print("------------")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Check missing values
print("MISSING VALUES")
print("--------------")
print(df.isnull().sum())

# Check missing values percentage
print("\nMISSING VALUE PERCENTAGE")
print("------------------------")
print((df.isnull().sum() / len(df) * 100).round(2))

# Check duplicate records
print("\nDUPLICATE RECORDS")
print("-----------------")
print("Total duplicates:", df.duplicated().sum())

# Check negative and zero quantities
print("\nQUANTITY CHECK")
print("--------------")
print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Zero quantities:", (df["Quantity"] == 0).sum())

# Check zero and negative prices
print("\nUNIT PRICE CHECK")
print("----------------")
print("Zero prices:", (df["UnitPrice"] == 0).sum())
print("Negative prices:", (df["UnitPrice"] < 0).sum())
```

```python
# Identify cancelled transactions

df["InvoiceNo"] = df["InvoiceNo"].astype(str)

cancelled = df[df["InvoiceNo"].str.startswith("C")]

print("CANCELLED TRANSACTIONS")
print("----------------------")
print("Cancelled rows:", len(cancelled))

print("\nSample cancelled transactions:")
display(cancelled.head(10))

print("\nQuantity summary:")
print(cancelled["Quantity"].describe())
```

```python
# Identify negative quantity transactions
negative_qty = df[df["Quantity"] < 0]

# Identify negative quantity transactions without C invoices
other_returns = negative_qty[
    ~negative_qty["InvoiceNo"].str.startswith("C")
]

print("TRANSACTION ANALYSIS")
print("--------------------")

print("Total negative quantity rows:", len(negative_qty))

print("Cancelled invoice rows:", len(cancelled))

print("Negative quantity rows without C invoice:",
      len(other_returns))

print("\nSample other return transactions:")
display(other_returns.head(10))

print("\nInvoice number examples:")
print(other_returns["InvoiceNo"].head(10).tolist())
```

```python
# Calculate transaction revenue
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print("REVENUE COLUMN CREATED")
print("----------------------")

print(df[["InvoiceNo", "Quantity", "UnitPrice", "Revenue"]].head(10))

print("\nTOTAL REVENUE INCLUDING RETURNS AND CANCELLATIONS:")
print(round(df["Revenue"].sum(), 2))
```

```python
# Create a separate DataFrame for valid sales
sales_df = df[
    (~df["InvoiceNo"].str.startswith("C")) &
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0)
].copy()

# Calculate revenue for valid sales
sales_df["Revenue"] = (
    sales_df["Quantity"] * sales_df["UnitPrice"]
)

print("VALID SALES ANALYSIS")
print("--------------------")

print("Original transaction rows:", len(df))
print("Valid sales rows:", len(sales_df))
print("Excluded rows:", len(df) - len(sales_df))

print("\nTOTAL VALID SALES REVENUE:")
print(round(sales_df["Revenue"].sum(), 2))

print("\nFIRST FIVE VALID SALES:")
display(sales_df.head())
```

```python
# Identify duplicate rows in the original dataset
duplicate_rows = df[df.duplicated(keep=False)].copy()

print("DUPLICATE ANALYSIS")
print("------------------")

print("Duplicate rows including original records:",
      len(duplicate_rows))

print("Duplicate groups (excluding first occurrence):",
      df.duplicated().sum())

print("\nSample duplicate records:")
display(
    duplicate_rows.sort_values(
        by=["InvoiceNo", "StockCode", "Quantity"]
    ).head(15)
)
```

```python
# Remove exact duplicate records from valid sales
sales_clean = sales_df.drop_duplicates().copy()

print("DUPLICATE CLEANING RESULTS")
print("--------------------------")

print("Rows before cleaning:", len(sales_df))
print("Rows after cleaning:", len(sales_clean))
print("Duplicate rows removed:", len(sales_df) - len(sales_clean))

print("\nRemaining duplicates:",
      sales_clean.duplicated().sum())

print("\nRevenue before cleaning:",
      round(sales_df["Revenue"].sum(), 2))

print("Revenue after cleaning:",
      round(sales_clean["Revenue"].sum(), 2))
```

```python
# Check missing values in cleaned sales data

print("MISSING VALUES ANALYSIS")
print("-----------------------")

missing_values = sales_clean.isnull().sum()

missing_percentage = (
    sales_clean.isnull().sum() / len(sales_clean)
) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage.round(2)
})

print(missing_summary)
```

```python
# Create dataset for customer-level analysis

customer_df = sales_clean.dropna(subset=["CustomerID"]).copy()

# Convert CustomerID to integer
customer_df["CustomerID"] = customer_df["CustomerID"].astype(int)

print("CUSTOMER DATASET RESULTS")
print("------------------------")

print("Total sales records:", len(sales_clean))
print("Customer records:", len(customer_df))
print("Records without CustomerID:",
      sales_clean["CustomerID"].isnull().sum())

print("Unique customers:",
      customer_df["CustomerID"].nunique())

print("Missing CustomerID after cleaning:",
      customer_df["CustomerID"].isnull().sum())
```

```python
# Check transaction date range

print("TRANSACTION DATE ANALYSIS")
print("-------------------------")

print("First transaction date:",
      sales_clean["InvoiceDate"].min())

print("Last transaction date:",
      sales_clean["InvoiceDate"].max())

print("Total days covered:",
      (sales_clean["InvoiceDate"].max()
       - sales_clean["InvoiceDate"].min()).days)

print("Total months covered:",
      sales_clean["InvoiceDate"].dt.to_period("M").nunique())

print("Total years covered:",
      sales_clean["InvoiceDate"].dt.year.nunique())
```

```python
# CUSTOMER PURCHASE FREQUENCY ANALYSIS

customer_orders = (
    customer_df.groupby("CustomerID")["InvoiceNo"]
    .nunique()
    .reset_index(name="Total Orders")
)

customer_orders["Customer Type"] = np.where(
    customer_orders["Total Orders"] > 1,
    "Repeat Customer",
    "One-Time Customer"
)

customer_summary = (
    customer_orders["Customer Type"]
    .value_counts()
    .reset_index()
)

customer_summary.columns = [
    "Customer Type",
    "Number of Customers"
]

customer_summary["Percentage"] = (
    customer_summary["Number of Customers"]
    / len(customer_orders)
) * 100

print("CUSTOMER BEHAVIOUR ANALYSIS")
print("--------------------------")

print(customer_summary.to_string(index=False))

print("\nTotal identified customers:",
      customer_orders["CustomerID"].nunique())
```

```python
# CUSTOMER REVENUE ANALYSIS

customer_revenue = (
    customer_df.groupby("CustomerID")["Revenue"]
    .sum()
    .reset_index()
)

customer_revenue.columns = [
    "CustomerID",
    "Total Revenue"
]

average_customer_revenue = customer_revenue[
    "Total Revenue"
].mean()

top_customers = customer_revenue.sort_values(
    by="Total Revenue",
    ascending=False
).head(10)

print("CUSTOMER REVENUE ANALYSIS")
print("-------------------------")

print("Total identified customers:",
      len(customer_revenue))

print("Average revenue per customer: £",
      round(average_customer_revenue, 2))

print("\nTOP 10 CUSTOMERS BY REVENUE")
print(top_customers.to_string(index=False))
```

```python
# RFM CUSTOMER ANALYSIS

# Set analysis date to one day after the last transaction
analysis_date = (
    customer_df["InvoiceDate"].max()
    + pd.Timedelta(days=1)
)

# Calculate RFM metrics
rfm = customer_df.groupby("CustomerID").agg(
    Recency=("InvoiceDate",
             lambda x: (analysis_date - x.max()).days),

    Frequency=("InvoiceNo", "nunique"),

    Monetary=("Revenue", "sum")
).reset_index()

print("RFM CUSTOMER ANALYSIS")
print("---------------------")

print("Analysis date:", analysis_date)

print("\nRFM dataset sample:")
print(rfm.head(10).to_string(index=False))

print("\nRFM summary:")
print(rfm[["Recency", "Frequency", "Monetary"]].describe().round(2))
```

```python
# RFM SCORING

rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

print("RFM SCORING RESULTS")
print("-------------------")

print(rfm.head(10).to_string(index=False))

print("\nRFM score distribution:")
print(rfm[["R_Score", "F_Score", "M_Score"]].describe())
```

```python
# CUSTOMER SEGMENTATION

def segment_customer(row):

    if row["RFM_Score"] == "555":
        return "Champions"

    elif row["R_Score"] >= 4 and row["F_Score"] >= 4:
        return "Loyal Customers"

    elif row["R_Score"] >= 4 and row["F_Score"] <= 2:
        return "New Customers"

    elif row["R_Score"] >= 3 and row["F_Score"] >= 2:
        return "Potential Loyalists"

    elif row["R_Score"] <= 2 and row["F_Score"] >= 3:
        return "At Risk"

    elif row["R_Score"] <= 2 and row["F_Score"] <= 2:
        return "Lost Customers"

    else:
        return "Needs Attention"


rfm["Customer_Segment"] = rfm.apply(
    segment_customer,
    axis=1
)

# Count customers in each segment
segment_summary = (
    rfm["Customer_Segment"]
    .value_counts()
    .reset_index()
)

segment_summary.columns = [
    "Customer Segment",
    "Number of Customers"
]

segment_summary["Percentage"] = (
    segment_summary["Number of Customers"]
    / len(rfm)
) * 100

print("CUSTOMER SEGMENTATION RESULTS")
print("-----------------------------")

print(segment_summary.round(2).to_string(index=False))

print("\nSample segmented customers:")
print(
    rfm[
        ["CustomerID", "R_Score", "F_Score",
         "M_Score", "RFM_Score", "Customer_Segment"]
    ].head(10).to_string(index=False)
)
```

```python
# CUSTOMER SEGMENT REVENUE ANALYSIS

# Merge customer segments with customer revenue
segment_revenue = rfm.merge(
    customer_revenue,
    on="CustomerID",
    how="inner"
)

# Calculate revenue by customer segment
segment_revenue_summary = (
    segment_revenue.groupby("Customer_Segment")
    .agg(
        Total_Revenue=("Total Revenue", "sum"),
        Number_of_Customers=("CustomerID", "nunique"),
        Average_Revenue_Per_Customer=("Total Revenue", "mean")
    )
    .reset_index()
)

# Calculate revenue contribution percentage
segment_revenue_summary["Revenue_Percentage"] = (
    segment_revenue_summary["Total_Revenue"]
    / segment_revenue_summary["Total_Revenue"].sum()
) * 100

# Sort by total revenue
segment_revenue_summary = segment_revenue_summary.sort_values(
    by="Total_Revenue",
    ascending=False
)

print("CUSTOMER SEGMENT REVENUE ANALYSIS")
print("---------------------------------")

print(
    segment_revenue_summary.round(2).to_string(index=False)
)
```

```python
# ORDER VALUE BY CUSTOMER SEGMENT

# Attach customer segments to individual orders
order_customer = (
    customer_df.groupby(["CustomerID", "InvoiceNo"])
    ["Revenue"].sum()
    .reset_index(name="Order_Revenue")
)

order_segment = order_customer.merge(
    rfm[["CustomerID", "Customer_Segment"]],
    on="CustomerID",
    how="inner"
)

# Calculate order metrics by segment
order_segment_summary = (
    order_segment.groupby("Customer_Segment")
    .agg(
        Total_Orders=("InvoiceNo", "nunique"),
        Average_Order_Value=("Order_Revenue", "mean"),
        Median_Order_Value=("Order_Revenue", "median"),
        Total_Revenue=("Order_Revenue", "sum")
    )
    .reset_index()
    .sort_values(
        by="Average_Order_Value",
        ascending=False
    )
)

print("ORDER VALUE BY CUSTOMER SEGMENT")
print("--------------------------------")

print(order_segment_summary.round(2).to_string(index=False))
```

```python
# AVERAGE ORDER VALUE BY CUSTOMER SEGMENT

aov_chart = order_segment_summary.sort_values(
    by="Average_Order_Value",
    ascending=False
)

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=aov_chart,
    x="Customer_Segment",
    y="Average_Order_Value"
)

ax.set_title(
    "Average Order Value by Customer Segment",
    fontsize=16,
    fontweight="bold",
    pad=20
)

ax.set_xlabel("Customer Segment")
ax.set_ylabel("Average Order Value (£)")

ax.yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f"£{x:,.0f}")
)

ax.bar_label(
    ax.containers[0],
    fmt="£%.2f",
    padding=4
)

plt.xticks(rotation=30, ha="right")
plt.tight_layout()

plt.savefig(
    "../outputs/average_order_value_by_segment.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
